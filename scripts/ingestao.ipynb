```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
import duckdb
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
# lendo o arquivo z0019_1.csv e adicionando as colunas nome_arquivo e data_ingestao
# criando um dataframe com os dados do arquivo z0019_1.csv

arquivo='z0019_1.csv'
data_ingestao = datetime.now()

df = pd.read_csv(f'../landing/{arquivo}' , sep=';')
df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestao
df.head()
```

```python
# criando a tabela bronze_produtos caso ela não exista

con.execute('''
    CREATE TABLE IF NOT EXISTS bronze_produtos (
        NATBR VARCHAR,
        MAKTX VARCHAR,
        WERKS VARCHAR,
        MAINS VARCHAR,
        LABST VARCHAR,
        nome_arquivo VARCHAR,
        data_ingestao timestamp       
    )
''')
```

```python
# lendo o arquivo z0019_2.csv e adicionando as colunas nome_arquivo e data_ingestao

resultado = con.execute('SELECT * FROM bronze_produtos').fetchdf()
resultado.head()
```

```python
# inserindo os dados do dataframe na tabela bronze_produtos

con.execute("insert into bronze_produtos select * from df")
```

```python
# lendo o arquivo z0019_2.csv e adicionando as colunas nome_arquivo e data_ingestao

arquivo='z0019_2.csv'
data_ingestao = datetime.now()

df = pd.read_csv(f'../landing/{arquivo}' , sep=';')
df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestao
df.head()
```

```python
# codigo não fara nada pois cira somente se nao existir a tabela, mas como ja existe, nao fara nada

con.execute('''
    CREATE TABLE IF NOT EXISTS bronze_produtos (
        NATBR VARCHAR,
        MAKTX VARCHAR,
        WERKS VARCHAR,
        MAINS VARCHAR,
        LABST VARCHAR,
        nome_arquivo VARCHAR,
        data_ingestao timestamp       
    )
''')
```

```python
# vou inserir os dados do arquivo z0019_2.csv na tabela bronze_produtos
con.execute("insert into bronze_produtos select * from df")
```

```python
# vou ler a tabela bronze_produtos para verificar se os dados foram inseridos corretamente
resultado = con.execute('SELECT * FROM bronze_produtos').fetchdf()
resultado.head(10)
```
